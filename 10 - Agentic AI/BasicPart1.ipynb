```python
def generate_pandas_code(user_instruction, df):
    schema_info = str(df.dtypes)

    prompt = f"""
        You are a senior data engineer.
        You are working with a pandas dataframe named df.

        Here is the schema: {schema_info}

        Rules: 
        - Only return valid python pandas code.
        - Do NOT explain anything.
        - Do NOT add markdown
        - Do NOT add backticks
        - The dataframe variable is already named df
        - The final result must be stored in a variable called result.

        User Request: {user_instruction}
    """

    response = client.chat.completions.create(
        model = 'gpt-4o-mini',
        temperature=0,
        messages = [{'role': 'user', 'content': prompt}]
    )

    return response.choices[0].message.content
```

```python
def execute_code(code, df):
    local_vars = {'df' : df.copy()}
    try:
        exec(code, {}, local_vars)
        return local_vars.get("result", "No result variable found.")
    except Exception as e:
        return f"Execution Error: {e}"
```

```python
def auto_schema_summary(df, max_examples=3):
    summary = []

    for col in df.columns:
        col_data = df[col]
        dtype = str(col_data.dtype)
        null_percent = round(col_data.isna().mean() * 100, 2)
        unique_values = col_data.nunique(dropna=True)

        column_info = {
            "column": col,
            "dtype": dtype,
            "null_percent": null_percent,
            "unique_values": int(unique_values)
        }

        if pd.api.types.is_numeric_dtype(col_data):
            column_info['min'] = col_data.min()
            column_info['max'] = col_data.max()
        else:
            examples = col_data.dropna().unique()[:max_examples]
            column_info['example_values'] = list(map(str, examples))

        summary.append(column_info)
    return summary
```

```python
def generate_pandas_code(user_instruction, df):
    columns = df.columns
    schema_info = auto_schema_summary(df)

    prompt = f"""
        You are a senior data engineer.
        You are working with a pandas dataframe named df.

        Here is the schema: {schema_info}
        List of columns: {list(df.columns)}

        Rules: 
        - Only return valid python pandas code.
        - Do NOT explain anything.
        - Do NOT add markdown
        - Do NOT add backticks
        - The dataframe variable is already named df
        - The final result must be stored in a variable called result.
        - You MUST use ONLY column names from the available columns list.
        - Do NOT invent or modify column names.
        - Column names are case-sensitive.
        - If unsure, choose the closest exact column from the list.
        - Only return valid Python pandas code.
        - Store final result in variable named result.        

        User Request: {user_instruction}
    """

    response = client.chat.completions.create(
        model = 'gpt-4o-mini',
        temperature=0,
        messages = [{'role': 'user', 'content': prompt}]
    )

    return response.choices[0].message.content
```

```python
import difflib

def suggest_column(user_input, columns):
    matches = difflib.get_close_matches(user_input, columns, n=3)
    return matches
```

```python
def correlation_summary(df, threshold = 0.7):
    numeric_df = df.select_dtypes(include = 'number')
    corr_matrix = numeric_df.corr()

    strong_corr = []
    for col in corr_matrix.columns:
        for idx in corr_matrix.index:
            if col != idx:
                val = corr_matrix.loc[idx, col]
                if abs(val) > threshold:
                    strong_corr.append((idx, col, round(val, 2)))
    return strong_corr
```
