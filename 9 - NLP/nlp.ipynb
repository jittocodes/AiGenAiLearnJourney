## Stemming

```python
words = ["eating", "eats", "eaten", "writing", "writes", "programming", "programs", 'history', 'congratulations']
```

```python
## Porter Stemmer
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word + " __ " + stemming.stem(word))
```

```python
# # history __ histori
# or
# congratulations __ congratul

# - Issue with PorterStemming
```

### RegexpStemmer Class

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
for word in words:
    print(word + " __ " + reg_stemmer.stem(word))
```

### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowball_stemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word + " __ " + snowball_stemmer.stem(word))
```

## Lemmatization

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemma = WordNetLemmatizer()
```

```python
for word in words:
    print(word + " __ " + lemma.lemmatize(word, pos='v'))
```

```python
lemma.lemmatize('goes')
```

```python
lemma.lemmatize('going', pos='v')
```

```python
lemma.lemmatize('going', pos='a')
```

```python
lemma.lemmatize('going', pos='r')
```

## NLTK

```python
import nltk
# nltk.download('punkt')
from nltk.tokenize import word_tokenize
```

```python
text = "Natural Language Processing is fun!"
nltk.download('punkt_tab')
tokens = word_tokenize(text)
print(tokens)
```

```python
text = [token.lower() for token in tokens]
```

```python
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
```

```python
filtered_tokens = [w for w in text if w.lower() not in stop_words]
print(filtered_tokens)
```

```python
## BOW and TF_IDF

from sklearn.feature_extraction.text import CountVectorizer
```

```python
courpus = ['I love NLP', 'I  love deep learning']
vectorizer = CountVectorizer()
x = vectorizer.fit_transform(courpus)
print(vectorizer.get_feature_names_out())
print(x.toarray())
```

```python
#TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(courpus)
print(vectorizer.get_feature_names_out())
print(x.toarray())
```
